```java
%%writefile ProcesoArbol.java
public class ProcesoArbol {

    public static void main(String[] args) throws Exception {

        // si no recibo argumento soy A (la raíz)
        String yo = (args.length > 0) ? args[0] : "A";

        long pid = ProcessHandle.current().pid();
        System.out.println("[" + yo + "] Proceso iniciado  | PID: " + pid);
        System.out.flush();

        // Hijos que debo crear según quién soy
        String[] hijos = hijosde(yo);

        // Crear los hijos concurrentemente (sin esperar entre uno y otro)
        Process[] procesos = new Process[hijos.length];
        for (int i = 0; i < hijos.length; i++) {
            procesos[i] = crearHijo(hijos[i]);
        }

        // Esperar a que todos terminen
        for (Process p : procesos) {
            p.waitFor();
        }

        // Pausa solo en A para que el docente pueda verificar el árbol
        if (yo.equals("A")) {
            System.out.println("\n=== Arbol completo creado. Presiona Enter para finalizar... ===");
            System.in.read();
        }

        System.out.println("[" + yo + "] Proceso finalizado | PID: " + pid);
        System.out.flush();
    }

     // Devuelve los hijos que le corresponden a cada nodo.
    static String[] hijosde(String nodo) {
        switch (nodo) {
            case "A": return new String[]{"B"};
            case "B": return new String[]{"C", "D"};
            case "C": return new String[]{"E", "F"};
            case "D": return new String[]{"G"};
            case "E": return new String[]{"H", "I"};
            default:  return new String[]{};  // F, G, H, I son hojas
        }
    }


     //Ejecuta java ProcesoArbol <hijo>
    static Process crearHijo(String nombreHijo) throws Exception {
        ProcessBuilder pb = new ProcessBuilder("java", "ProcesoArbol", nombreHijo);
        pb.inheritIO();  // el hijo imprime en la misma consola
        return pb.start();
    }
}
```
